```python
import pandas as pd
import numpy as np
import torch

# 예시: seaborn 내장 데이터 (인터넷 연결 필요)
import seaborn as sns
df = sns.load_dataset("penguins")
df.head()
```

## 문제 1. 데이터 구조 확인 (20점)
`shape`, `columns`, `dtypes`, `head()`, `info()` 를 출력하고 무엇을 알 수 있었는지 한 줄로 적으세요.


```python
# 답안

# 데이터 구조 출력
print("행과 열 개수:", df.shape)
print("열 이름:", df.columns)
print("자료형:", df.dtypes)
print("앞의 5행:", df.head())

df.info()

# 전처리 전 데이터 복사
before = df.copy()

# 펭귄의 종류, 서식 섬, 신체 측정값, 성별로 구성되어 있으며
# 숫자형 열과 문자형 열이 함께 있음을 확인했다.
```

## 문제 2. 결측값 처리 (25점)
열별 결측 개수를 출력하고, 각 열을 **삭제 또는 대체** 중 무엇으로 처리했는지 이유와 함께 적으세요.


```python
# 답안

# 처리 전 열별 결측 개수
print("처리 전 결측값:\n", df.isnull().sum())

# 신체 측정값: 임의의 값으로 대체하지 않고 실제 측정값이 있는 행만 사용
# 성별: 다른 성별로 임의 대체하지 않고 성별이 확인된 행만 사용
# species와 island도 결측값이 있다면 같은 기준으로 해당 행을 삭제
clean = df.dropna().copy()

# 처리 후 열별 결측 개수
print("처리 후 결측값:\n", clean.isnull().sum())
```

## 문제 3. 자료형 처리 (25점)
문자형 열을 숫자로 바꾸세요. 값이 두 개면 0/1, 셋 이상이면 원-핫 인코딩을 씁니다.


```python
# 답안

# 성별을 0과 1로 변환
clean["sex"] = clean["sex"].map({
    "Male": 0,
    "Female": 1
})

# 펭귄 종류와 섬을 원-핫 인코딩
clean = pd.get_dummies(
    clean,
    columns=["species", "island"],
    dtype=int
)

print(clean.head())
print("변환 후 자료형:\n", clean.dtypes)
```

## 문제 4. 전처리 전후 비교표 (20점)
아래 형태의 표를 만들어 출력하세요.

| 항목 | 전처리 전 | 전처리 후 |
| --- | --- | --- |
| 행 개수 | | |
| 열 개수 | | |
| 결측 개수 | | |
| object 열 개수 | | |


```python
# 답안 — pd.DataFrame 으로 만들어 출력하세요

# 전처리 전후 상태를 표로 정리
table = pd.DataFrame({
    "항목": [
        "행 개수",
        "열 개수",
        "결측 개수",
        "object 열 개수"
    ],
    "전처리 전": [
        before.shape[0],
        before.shape[1],
        before.isnull().sum().sum(),
        (before.dtypes == "object").sum()
    ],
    "전처리 후": [
        clean.shape[0],
        clean.shape[1],
        clean.isnull().sum().sum(),
        (clean.dtypes == "object").sum()
    ]
})

print(table.to_string(index=False))
```

## 문제 5. Tensor 변환 (10점)
전처리한 입력을 `float32` Tensor로 바꾸고, 개수 · shape · dtype · 최솟값 · 최댓값 · NaN 개수를 출력하세요.


```python
# 답안

# 전처리한 데이터를 float32 NumPy 배열로 변환
X_np = clean.to_numpy(dtype="float32")

# NumPy 배열을 Tensor로 변환
X = torch.from_numpy(X_np)

# PPT 108쪽의 확인 항목 출력
print("데이터 개수:", X.shape[0])
print("shape:", tuple(X.shape))
print("dtype:", X.dtype)
print("최솟값:", X.min().item())
print("최댓값:", X.max().item())
print("NaN 개수:", torch.isnan(X).sum().item())
```
