## 문제 1. 숫자 5개의 평균과 최댓값 (20점)
리스트에 숫자 5개를 담고, 평균과 최댓값을 출력하세요. 평균은 소수점 둘째 자리까지.


```python
# 답안

# 숫자 5개를 리스트에 저장
nums = [10, 20, 30, 40, 50]

# 평균과 최댓값 계산
mean = sum(nums) / len(nums)
best = max(nums)

print("평균" , mean)
print("최댓값:", best)
```

## 문제 2. 하루 지출 계산 프로그램 (40점)
항목 이름과 금액을 딕셔너리에 저장하고 아래를 출력하세요.
- 총 지출액
- 가장 많이 쓴 항목과 금액
- 항목별 지출이 총액에서 차지하는 비율(%)


```python
expenses = {
    "점심": 9000,
    "커피": 4500,
    "교통": 2800,
    "간식": 3200,
}

# 답안

# 총 지출액
total = sum(expenses.values())
print("총 지출액:", total, "원")

# 가장 많이 쓴 항목과 금액
most = max(expenses, key=expenses.get)
print("가장 많이 쓴 항목:", most, expenses[most], "원")

# 항목별 지출 비율
for name, amount in expenses.items():
    ratio = amount / total * 100
    print(f"{name}: {ratio:.2f}%")
```

## 문제 3. 오류 고치기 (20점)
아래 셀에는 오류가 있습니다. 실행해 메시지를 읽고 고친 뒤, **어떤 오류였는지** 주석으로 적으세요.


```python
data = [3, 7, 11]
result = 0
#data 는 0-2 까지 이다.
for i in range(len(data)):
    result += data[i]
# 문자열 끼리는 합할 수 없다.
print("합계:", result)
```

```python
data = [3, 7, 11]
result = 0
for i in range(4):
    result += data[i]
print("합계: " + result)
```

## 문제 4. Tensor 만들기 (20점)
1. `torch.arange` 로 0부터 11까지의 텐서를 만드세요.
2. 그것을 (3, 4) 모양으로 바꾸세요.
3. shape, dtype, 전체 평균을 출력하세요.


```python
import torch

# 답안

# 0부터 11까지 생성
x = torch.arange(0, 12, dtype=torch.float32)

# (3, 4) 모양으로 변경
x = x.reshape(3, 4)

print("shape:", x.shape)
print("dtype:", x.dtype)
print("전체 평균:", x.mean())
```
