## 문제 1. Tensor 5개 생성과 속성 확인표 (40점)
서로 다른 Tensor 5개를 만들고, 아래 항목을 담은 표를 출력하세요.

| 이름 | shape | dim | dtype | device |

각 Tensor를 왜 그렇게 만들었는지 주석으로 한 줄 적으세요.


```python
# 답안

# 라이브러리
import torch
import pandas as pd

# 세 개의 실수 데이터를 하나의 Tensor에 저장하기 위해 생성
a = torch.tensor([1.0, 2.0, 3.0])

# 계산 결과를 저장할 공간을 0으로 초기화하기 위해 생성
b = torch.zeros(2, 3)

# 모든 원소가 1인 Tensor의 생성 방법을 확인하기 위해 생성
c = torch.ones(2, 3)

# 일정한 간격의 정수 데이터를 만들기 위해 생성
d = torch.arange(0, 10, 2)

# 난수로 값을 생성하기 위해 생성
e = torch.randn(2, 3)




# 각 Tensor의 속성을 표로 정리
table = pd.DataFrame([
    ["a", tuple(a.shape), a.dim(), str(a.dtype), str(a.device)],
    ["b", tuple(b.shape), b.dim(), str(b.dtype), str(b.device)],
    ["c", tuple(c.shape), c.dim(), str(c.dtype), str(c.device)],
    ["d", tuple(d.shape), d.dim(), str(d.dtype), str(d.device)],
    ["e", tuple(e.shape), e.dim(), str(e.dtype), str(e.device)]
], columns=["이름", "shape", "dim", "dtype", "device"])

# 표 출력
print(table.to_string(index=False))
```

## 문제 2. Tensor 연산 (20점)

## 문제 3. CPU와 GPU 비교 (20점)
같은 크기의 행렬곱을 CPU와 GPU에서 각각 실행해 걸린 시간을 출력하고, 차이를 한두 줄로 설명하세요.
GPU를 쓸 수 없는 경우에는 CPU 결과와 그 이유를 적으세요.


```python
# 답안

# 라이브러리
import torch
import time

# 행렬곱 평균 실행 시간 측정
def bench(device, n=5000, repeat=3):
    a = torch.randn(n, n, device=device)
    b = torch.randn(n, n, device=device)

    if device == "cuda":
        torch.cuda.synchronize()

    start = time.time()

    for _ in range(repeat):
        a @ b

    if device == "cuda":
        torch.cuda.synchronize()

    return (time.time() - start) / repeat

# CPU 시간 출력
cpu_time = bench("cpu")
print(f"CPU: {cpu_time:.4f}초")

# GPU 시간 출력
if torch.cuda.is_available():
    gpu_time = bench("cuda")
    print(f"GPU: {gpu_time:.4f}초")


print('같은 크기의 행렬곱에서 GPU의 실행 시간이 CPU보다 짧았다. 행렬곱에서는 GPU가 유리하다.')
```

## 문제 4. 자동미분 (20점)
아래 두 가지를 하세요.
1. $y = 2x^3 + 5x$ 에서 $x=3$ 일 때의 도함수를 자동미분으로 구하고, 손으로 구한 값과 비교하세요.
2. 자기가 만든 수식 하나에 대해 미분값을 구하고, 손 계산과 일치하는지 확인하세요.


```python
# 답안

# 라이브러리
import torch

# 1. y = 2x^3+5x에서 x=3일 때 자동미분
x = torch.tensor(3.0, requires_grad=True)
y = 2*x**3+5*x

y.backward()
print("1번 자동미분:",x.grad)

print("1번 손 계산:", 6*3**2+5)


# 2. 직접 만든 수식: y =2x^3+5x에서 x=2일 때
x = torch.tensor(2.0, requires_grad=True)
y = 2*x**3+5*x

y.backward()
print("2번 자동미분:", x.grad)

print("2번 손 계산:", 6*x**2+5)
```
