## 문제 1. 5×5 숫자 배열 생성 (20점)
0부터 100 사이의 정수로 5×5 배열을 만들고, shape · ndim · dtype 을 출력하세요.
(힌트: `np.random.randint`, 결과를 고정하려면 `np.random.seed(0)`)


```python
# 답안

np.random.seed(0)

# 0부터 100까지의 정수로 5×5 배열 생성
a = np.random.randint(0, 101, size=(5, 5))

print(a)
print("shape:", a.shape)
print("ndim:", a.ndim)
print("dtype:", a.dtype)
```

## 문제 2. 최댓값의 위치 찾기 (30점)
문제 1의 배열에서 아래를 출력하세요.
- 최댓값과 그 (행, 열) 위치
- 행별 평균, 열별 평균


```python
# 답안

# 최댓값의 위치를 (행, 열)로 변환
pos = np.unravel_index(a.argmax(), a.shape)

print("최댓값:", a.max())
print("최댓값 위치 (행, 열):", pos)

# axis=1: 각 행의 평균
print("행별 평균:", a.mean(axis=1))

# axis=0: 각 열의 평균
print("열별 평균:", a.mean(axis=0))
```

## 문제 3. 흑백 이미지처럼 시각화 (30점)
문제 1의 배열을 `imshow(cmap="gray")` 로 그리고, 최댓값 위치의 픽셀을 255로 바꿔 다시 그리세요.
두 그림을 나란히 배치하세요. (힌트: `plt.subplots(1, 2)`)


```python
# 답안

# 원본을 복사
changed = a.copy()

# 최댓값 위치의 픽셀을 255로 변경
changed[pos] = 255

# 두 그림을 나란히 배치
fig, axes = plt.subplots(1, 2)

axes[0].imshow(a, cmap="gray", vmin=0, vmax=255)
axes[1].imshow(changed, cmap="gray", vmin=0, vmax=255)

plt.show()
```

## 문제 4. 오류 고치기 (20점)
아래 셀에는 오류가 있습니다. 실행해 메시지를 읽고 고친 뒤, **어떤 오류였고 shape이 어떻게 어긋났는지** 주석으로 적으세요.


```python
p = np.arange(12).reshape(3, 4)
q = np.arange(12).reshape(3, 4)

# 행렬곱에서 p의 열 수 4와 q의 행 수 3이 달라 오류 발생
# q.T로 q의 shape을 (4, 3)으로 바꾸어 가운데 차원을 맞춤
# (3, 4) @ (4, 3) → (3, 3)
print(p @ q.T)
```

```python
p = np.arange(12).reshape(3, 4)
q = np.arange(12).reshape(3, 4)

print(p @ q)
```
